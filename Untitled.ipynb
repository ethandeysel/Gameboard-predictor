```python
# ==========================================
# CELL 1: Magic Commands & Environment Setup
# ==========================================
%load_ext autoreload
%autoreload 2

import os
# Force single-threading for PyTorch/Linear Algebra to prevent notebook lag
os.environ["MKL_NUM_THREADS"] = "1"
os.environ["OPENBLAS_NUM_THREADS"] = "1"
os.environ["OMP_NUM_THREADS"] = "1"

import numpy as np
import pandas as pd
import torch
import gpytorch

# Import your modular repository assets
from src.data.database import get_history_df
from src.models.GP import SpatiotemporalGPModel, build_and_train_gp, predict_next_week

print("Environment and modules successfully loaded.")
```

```python
# ==========================================
# CELL 2: Walk-Forward Data Masking
# ==========================================
print("Fetching chronological history from database...")
df = get_history_df()
df_clean = df.dropna(subset=['reward']).copy()

# Define the target validation week
test_week = 19

# Split data: Train on everything BEFORE Week 19, test ON Week 19
train_df = df_clean[df_clean['week_number'] < test_week].copy()
test_df = df_clean[df_clean['week_number'] == test_week].copy()

# Ensure the test grid indices match up sequentially from 0 to 19
test_df = test_df.sort_values('grid_index')

print(f"Training Data: Weeks 1 to {test_week-1} ({len(train_df)} total observations)")
print(f"Validation Data: Week {test_week} ({len(test_df)} total observations)")

# Transform spatial and target data into the design matrix
train_df['x_coord'] = train_df['grid_index'] % 4
train_df['y_coord'] = train_df['grid_index'] // 4
train_df['log_reward'] = np.log(train_df['reward'].astype(float) + 1e-6)

X_train = train_df[['x_coord', 'y_coord', 'week_number']].values
y_train = train_df['log_reward'].values
```

```python
# ==========================================
# CELL 3: Retrain GPyTorch Model
# ==========================================
print(f"--- Retraining Model (Blind to Week {test_week}) ---")

# Run the training pipeline (GPyTorch optimizes hyperparameters via Adam)
model, likelihood = build_and_train_gp(X_train, y_train, training_iterations=150)

# Generate predictions explicitly for the held-out week
best_tile_pred, evs, ucbs = predict_next_week(model, likelihood, target_week=test_week)
```

```python
# ==========================================
# CELL 4: Tactical Grading & Heatmaps
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

actual_rewards = test_df['reward'].values
actual_best_tile = np.argmax(actual_rewards)

print("\n" + "="*42)
print(f"🎯 WEEK {test_week} BACKTEST ANALYSIS REPORT 🎯")
print("="*42)

print(f"\n[THE MODEL'S RECOMMENDED TARGET]")
print(f"Predicted Best Tile     : Tile {best_tile_pred} (Col {best_tile_pred % 4}, Row {best_tile_pred // 4})")
print(f"Model Expected Value (EV): {evs[best_tile_pred]:.1f} points")
print(f"Model Confidence (UCB)  : {ucbs[best_tile_pred]:.1f} points")
print(f"Actual Payout Received  : {actual_rewards[best_tile_pred]:.1f} points")

print(f"\n[THE TRUE JACKPOT TARGET]")
print(f"True Maximum Tile       : Tile {actual_best_tile} (Col {actual_best_tile % 4}, Row {actual_best_tile // 4})")
print(f"True Payout Received    : {actual_rewards[actual_best_tile]:.1f} points")
print(f"Model's EV for this Tile: {evs[actual_best_tile]:.1f} points")

# Calculate Global Accuracy
mae = np.mean(np.abs(evs - actual_rewards))
print(f"\n[BOARD-WIDE ACCURACY]")
print(f"Mean Absolute Error (MAE): {mae:.1f} points deviation per tile")

print(f"\n[TOP 5 PREDICTED HIGHEST-VALUE TARGETS]")
print("-" * 65)
print(f"{'Rank':<5} | {'Tile':<5} | {'UCB Score':<12} | {'Expected EV':<12} | {'Actual Payout':<13}")
print("-" * 65)

top_5_indices = np.argsort(ucbs)[::-1][:5]
for rank, idx in enumerate(top_5_indices, start=1):
    # FIXED: The format specifier is now safely just <5
    print(f"{rank:<5} | {idx:<5} | {ucbs[idx]:<12.1f} | {evs[idx]:<12.1f} | {actual_rewards[idx]:<13.1f}")
print("-" * 65)

# ==========================================
# 5. Visual Heatmap Generation
# ==========================================
# Reshape the 1D arrays into 5x4 grids
ev_matrix = evs.reshape(5, 4)
ucb_matrix = ucbs.reshape(5, 4)
actual_matrix = actual_rewards.reshape(5, 4)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Heatmap 1: Expected Value
sns.heatmap(ev_matrix, annot=True, fmt=".0f", cmap="viridis", ax=axes[0], cbar_kws={'label': 'EV Points'})
axes[0].set_title(f"Model Expected Value (Week {test_week})", fontsize=14)
axes[0].set_xlabel("Grid Column (X)")
axes[0].set_ylabel("Grid Row (Y)")

# Heatmap 2: Upper Confidence Bound (The Decision Maker)
sns.heatmap(ucb_matrix, annot=True, fmt=".0f", cmap="magma", ax=axes[1], cbar_kws={'label': 'UCB Points'})
axes[1].set_title(f"Upper Confidence Bound (Optimism)", fontsize=14)
axes[1].set_xlabel("Grid Column (X)")

# Highlight the model's chosen tile in Cyan
best_y, best_x = best_tile_pred // 4, best_tile_pred % 4
axes[1].add_patch(plt.Rectangle((best_x, best_y), 1, 1, fill=False, edgecolor='cyan', lw=4))

# Heatmap 3: The Actual Reality
sns.heatmap(actual_matrix, annot=True, fmt=".0f", cmap="plasma", ax=axes[2], cbar_kws={'label': 'Actual Points'})
axes[2].set_title(f"Actual Reality (Week {test_week})", fontsize=14)
axes[2].set_xlabel("Grid Column (X)")

# Highlight the model's chosen tile on the reality board to see if it hit
axes[2].add_patch(plt.Rectangle((best_x, best_y), 1, 1, fill=False, edgecolor='cyan', lw=4))

plt.tight_layout()
plt.show()
```

```python
# Add this to your Jupyter notebook after training
def print_kernel_diagnostics(model):
    # Extract the lengthscales from the kernel
    spatial_ls = model.covar_module.base_kernel.kernels[0].lengthscale.detach().numpy()
    temporal_ls = model.covar_module.base_kernel.kernels[1].lengthscale.detach().numpy()
    
    print(f"--- Spatial Reach (Lengthscale) ---")
    print(f"Spatial reach: {spatial_ls[0][0]:.2f} units")
    print(f"Temporal reach: {temporal_ls[0][0]:.2f} weeks")
    
    if spatial_ls[0][0] > 0.5:
        print("✅ Hot zones are 'bleeding' into neighbors (Spatial structure detected).")
    else:
        print("⚠️ Model is treating tiles as independent. Increase spatial prior.")

print_kernel_diagnostics(model)
```

```python
# Compare GP Mean vs 3-week Moving Average
df = get_history_df()
pivot_table = df.pivot(index='week_number', columns='grid_index', values='reward')
moving_avg = pivot_table.rolling(window=3).mean().iloc[-1].values

# Get GP predictions
_, gp_ev, _ = predict_next_week(model, likelihood, target_week=20)

# Check correlation
correlation = np.corrcoef(moving_avg, gp_ev)[0, 1]
print(f"GP vs. Moving Average correlation: {correlation:.2f}")

if correlation < 0.6:
    print("🚀 GP is finding complex patterns the simple moving average missed.")
else:
    print("GP is mostly following the simple moving average.")
```

```python
# ==========================================
# CELL: Define the X_new Grid
# ==========================================
# We define the coordinates for the current prediction week
target_week = int(df_clean['week_number'].max() + 1)

X_new = []
for tile_idx in range(20):
    x_coord, y_coord = tile_idx % 4, tile_idx // 4
    X_new.append([x_coord, y_coord, target_week])

X_new = np.array(X_new, dtype=float)
print(f"X_new grid initialized for Week {target_week}.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get Variance
model.eval()
test_x = torch.tensor(X_new, dtype=torch.float32)
with torch.no_grad():
    pred = likelihood(model(test_x))
    var = pred.variance.numpy().reshape(5, 4)

plt.figure(figsize=(8, 6))
sns.heatmap(var, annot=True, cmap="YlOrRd_r", cbar_kws={'label': 'Model Uncertainty (Variance)'})
plt.title("Certainty Map (Darker = More Certain of Hot Zone)")
plt.show()
```
